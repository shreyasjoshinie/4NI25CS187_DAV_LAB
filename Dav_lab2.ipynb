```python
import pandas as pd
import numpy as np
```

```python
data={
    "Name":["Amit","Bedo","Chetan","Amit","Divya","Ravi"],
    "Age":[25,np.nan,30,25,29,22],
    "Marks":[85,90,np.nan,85,78,95],
    "City":["Mysuru","Mysuru","Beng","Myusur",np.nan,"Delhi"]
}
df = pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\nShape of dataset (row,column):",df.shape)

print("\nColumn names:",df.columns.tolist())

print("\nData data of each coulumn:")
print(df.dtypes)
print("\nDataset info:")
df.info()
```

```python
print("\nDescriptive Statistics (numerical column):")
print(df.describe())

print("\nMean\n",df.mean(numeric_only=True))
print("\nMedian\n",df.median(numeric_only=True))
print("\nMode\n",df.mode(numeric_only=True))
print("\nStandard Deviation\n",df.std(numeric_only=True))
```

```python
print("\nMissing value in each column: ")
print(df.isnull().sum())
print("\nTotal missing values in dataset:",df.isnull().sum().sum())
```

```python
print("\nDuplicate row in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicate rows:",df.duplicated().sum())
print("\nEDA process complete successfully")
```
