```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("/content/student_performance_dataset.csv")
print("First 5 row of dataset:")
print(df.head())
```

```python
print("\nMissing values before handling")
print(df.isnull().sum())
```

```python
df.loc[0:5,"Study_Hours_per_Week"]=None
df.loc[0:5,"Attendance_Rate"]=None
df.loc[0:5,"Past_Exam_Scores"]=None
df.loc[0:5,"Final_Exam_Scores"]=None

print("\nMissing values after inserting sample Marks")
print(df.isnull().sum()[df.isnull().sum()>0])
```

```python
df["Study_Hours_per_Week"]=df["Study_Hours_per_Week"].fillna(df["Study_Hours_per_Week"].mean())
df["Attendance_Rate"]=df["Attendance_Rate"].fillna(df["Attendance_Rate"].median())
```

```python
df["Final_Exam_Score"]=df["Final_Exam_Score"].fillna(df["Final_Exam_Score"].mean())
print("\nMissing values after inpulation:")
print(df.isnull().sum())
```

```python
df=pd.get_dummies(df,columns=["Parental_Education_Level"])
print("\nOne-Hot encoded columns")
print(df.columns.tolist())
```

```python
ls=LabelEncoder()
df["Pass_Fail_Encoded"]=ls.fit_transform(df["Pass_Fail"])
print("\nLabl Encoded result")
print(df[["Pass_Fail","Pass_Fail_Encoded"]].head())
```

```python
df["Gender_Encoded"]=ls.fit_transform(df["Gender"])
df["Internet_Access_Encoded"]=ls.fit_transform(df["Internet_Access_at_Home"])
df["Extracurricular_Encoded"]=ls.fit_transform(df["Extracurricular_Activities"])
```

```python
df.to_csv("preprocessed_data.csv",index=False)
print("\nCleaned dataset saved as student_performance_cleaned.csv")
```
